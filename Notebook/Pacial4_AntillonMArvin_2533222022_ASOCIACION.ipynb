**Cargar librerías**



```python
import pandas as pd
import numpy as np

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

**Cargar el CSV**



```python
url = "https://raw.githubusercontent.com/MarvinAntillon97/Parcial4_AntillonMarvin_2533222022/refs/heads/main/Archivos/clave_E_asociacion.csv"

df = pd.read_csv(url)

df.head()
```

**Estructura del dataset**


```python
print(df.shape)

print(df.columns)

print(df.info())
```

```python
print(df.isnull().sum())

print(df.duplicated().sum())
```

```python
df.head()
```

# **Crear matriz de transacciones**


```python
basket = (
    df.groupby(['transaccion_id', 'item'])['cantidad']
    .sum()
    .unstack()
    .fillna(0)
)

basket.head()
```

# **Convertir cantidades a booleanos**



```python
basket_sets = basket.applymap(lambda x: 1 if x > 0 else 0)

basket_sets.head()
```

# **Aplicar algoritmo Apriori**



```python
frequent_items = apriori(
    basket_sets,
    min_support=0.05,
    use_colnames=True
)

frequent_items.sort_values(by='support', ascending=False).head(10)
```

# **Generar reglas de asociación**




```python
rules = association_rules(
    frequent_items,
    metric="lift",
    min_threshold=1
)

rules.head()
```

**Mostrar las 10 reglas más relevantes**


```python
top_rules = rules.sort_values(
    by='lift',
    ascending=False
)

top_rules[
    [
        'antecedents',
        'consequents',
        'support',
        'confidence',
        'lift'
    ]
].head(10)
```
